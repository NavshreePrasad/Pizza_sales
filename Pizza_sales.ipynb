## KPI's

```python
total_revenue= df['total_price'].sum()
print(f"total_revenue:${total_revenue:.2f}")
```

```python
total_pizzas_sold=df['quantity'].sum()
print("total_pizzas_sold: ", total_pizzas_sold)
```

```python
total_orders= df['order_id'].nunique()
print ("total_orders: ", total_orders)
```

```python
average_order_value= total_revenue/total_orders
print (f"average_order_value:${average_order_value:.2f}")
```

```python
average_pizzas_per_order= total_pizzas_sold/total_orders
print(f'average_pizzas_per_order: {average_pizzas_per_order:.2f}')
```
